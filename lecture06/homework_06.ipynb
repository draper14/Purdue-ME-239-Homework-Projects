+ Write a Python function that takes a list as an argument and returns the number of elements that are greater than a given element (also passed as an argument to the function). Call that function ``find_number_of_elms_greater_than(a, elm)``. Make sure you follow best practices when writing the docstring of your function.

````python
# your code here
def find_number_of_elms_greater_than(a, elm):
  """
  returns the number of elements in list "a" that are greater than element "elm"
  Inputs are list a, int elm
  Outputs int
  """
  count=0
  for i in a:
    if i>elm:
      count+=1
  return count
````

## Problem 2 - High-performance buildings revisited

In this problem we will continue analyzing the high-performance buildings dataset we introduced in {ref}`lecture03:homework:problem1` and with which we played in {ref}`lecture06:selecting-rows`.
Let me set you up by downloading and cleaning the data file:

````python
!curl -O 'https://raw.githubusercontent.com/PurdueMechanicalEngineering/me-239-intro-to-data-science/master/data/temperature_raw.xlsx'
import pandas as pd

df = pd.read_excel('temperature_raw.xlsx')
df = df.dropna(axis=0)
df.date = pd.to_datetime(df['date'], format='%Y-%m-%d')
df.head()
````

+ Plot the external temperature `t_out`

````python
# Your code here
plt.figure(figsize=(12, 4))
plt.xticks(rotation=90)

plt.scatter(df["household"],df["t_out"])
plt.xlabel("household")
plt.ylabel("t_out")
plt.title("household vs. t_out")
````

+ Extract the data pertaining to household `a5`.
Put the result in a new dataframe called `df_a5`.

````python
df_a5 = df[df['household'] == 'a5'].round(2)
df_a5.head()
````

+ For household `a5`, plot `t_unit` as a function of date.

````python
plt.xticks(rotation=90)
plt.scatter(df_a5["date"],df_a5["t_unit"])
plt.xlabel("date")
plt.ylabel("t_unit")
plt.title("date vs. t_unit of a5")
````

+ In a single figure, plot `date` vs `t_unit` for households `a5` and `a11`.

````python
df_new = df[(df['household'] == 'a5') | (df['household'] == 'a11')].round(2)
plt.xticks(rotation=90)
plt.scatter(df_new["date"],df_new["t_unit"])
plt.xlabel("date")
plt.ylabel("t_unit")
plt.title("date vs. t_unit of a5&a11")
````

+ In the same figure, plot the `t_out` and `t_unit` scatter plots for both households `a5` and `a11`.

````python
plt.scatter(df_new["t_out"],df_new["t_unit"])
plt.xlabel("t_out")
plt.ylabel("t_unit")
plt.title("date vs. t_unit of a5&a11")
````

+ In the same figure, plot the `t_out` and `hvac` scatter plots for both households `a5` and `a11`.

````python
plt.scatter(df_new["t_out"],df_new["hvac"])
plt.xlabel("t_out")
plt.ylabel("hvac")
plt.title("t_out vs. hvac of a5&a11")
````

+ In the same figure, plot the histogram of `t_unit` for households `a5` and `a11`.
Which household prefers cooler temperatures?
Hint: To make the histogram more appealing use the keywords ``density=True, alpha=0.25``.

````python
df_a5=df[df['household'] == 'a5']
df_a11=df[df['household'] == 'a11']
plt.hist(df_a5['t_unit'],alpha=0.25,density=True,label='a5')
plt.hist(df_a11['t_unit'],alpha=0.25,density=True,label='a11')
plt.legend()
plt.xlabel('t_unit')
plt.ylabel("frequency")
plt.title("t_unit histogram for a5&a11")
````

A11 has higher frequencies of lower values of t_unit, so A11 households prefer colder temperatures.

+ In the same figure, plot the histogram of `hvac` for households `a5` and `a11`.
Which household is more energy efficient (if any) and why?

````python
plt.hist(df_a5['hvac'],alpha=0.25,density=True,label='a5')
plt.hist(df_a11['hvac'],alpha=0.25,density=True,label='a11')
plt.legend()
plt.xlabel('hvac')
plt.ylabel("frequency")
plt.title("hvac histogram for a5&a11")
````

I can tell from the graphs that a5 has a lower average hvac energy due to the a11 outliers, corresponding to a more efficient household.

+ Repeat the analysis above for households `b17` and `c40`. Which household prefers cooler temperatures and which one is more energy efficient?

````python
df_b17=df[df['household'] == 'b17']
df_c40=df[df['household'] == 'c40']
fig, axes = plt.subplots(2, 1)
axes[0].hist(df_b17['hvac'],alpha=0.25,density=True,label='b17')
axes[0].hist(df_c40['hvac'],alpha=0.25,density=True,label='c40')
axes[0].legend()
axes[0].set_xlabel('hvac')
axes[0].set_ylabel("frequency")
axes[0].set_title("hvac histogram for b17&c40")

axes[1].hist(df_b17['t_unit'],alpha=0.25,density=True,label='b17')
axes[1].hist(df_c40['t_unit'],alpha=0.25,density=True,label='c40')
axes[1].legend()
axes[1].set_xlabel('t_unit')
axes[1].set_ylabel("frequency")
axes[1].set_title("t_unit histogram for b17&c40")

plt.tight_layout()# so they don't bunch up
````
